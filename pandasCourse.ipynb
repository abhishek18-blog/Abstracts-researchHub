```python
import pandas as pd

"""
A Pandas Series is a one-dimensional labeled array capable of holding data
of any type, such as integers, floats, strings, and Python objects.
You can think of it as a single column in an Excel spreadsheet or a database table
"""

num = pd.Series([
    10,20,30,40
])

print(num)
# left is index
# right is value
```

```python
# CREATING A SERIES FROM DIFFERENT DATA TYPES
# FROM A LIST
marks = pd.Series([80, 90, 75, 88])

# FROM A TUPLE
marks = pd.Series((80, 90, 75))

# FROM A NUMPY ARRAY
import numpy as np
arr = np.array([5, 10, 15])
series = pd.Series(arr)


# From a Dictionary
student = {
    "Rahul": 90,
    "Priya": 95,
    "Amit": 85
}

series = pd.Series(student)

print(series)
```

## Resources

| Data Structure | Ordered? | Mutable (Changeable)? | Element Types | Syntax | Best Used For |
|---|---|---|---|---|---|
| List | Yes | Yes | Mixed (Any type) | `[1, "apple", 3.4]` | Storing a flexible collection of item sequences |
| Tuple | Yes | No (Immutable) | Mixed (Any type) | `(1, "apple", 3.4)` | Protecting data from accidental changes |
| NumPy Array | Yes | Yes | Same type only | `np.array([1, 2, 3])` | Fast, heavy mathematical and numeric operations |
| Dictionary | Yes* | Yes | Key-Value pairs | `{"key": "value"}` | Fast lookups using unique labels or metadata |

```python
# EXERCISE
# 1. Create a series containing five numbers
nums = pd.Series([10, 20, 30, 40, 50])
print(nums)

# 2. Create a series from a tuple
nums2 = pd.Series((
    10, 20, 30, 40, 50
))

# 3. Create a series from NumPy array
nums3 = np.array([
    10,20,30,40
])
series = pd.Series(nums3)

# 4. Create a series with custom
marks = pd.Series(
    [20,30,40],
    index = ["Rahul","Priya","Amit"]
)

# access third element by its index
print(marks.iloc[0]) #use iloc to locate the index

# create series from a dictionary and print the value for one key
dictionary = {
    "Car" :5,
    "Bike" :8,
    "Bus" :2
}
print(dictionary["Car"])

"""
print(marks.dtype)
print(marks.shape)
print(marks.size)
print(marks.ndim)
print(marks.index)
print(marks.values)
"""

# ARITHMETIC BUILT-IN METHOD
print(marks.add(10))
print(marks.sub(10))
print(marks.mul(10))
print(marks.div(10))

# STATISTICAL BUILT-IN METHOD
print(marks.mean())
print(marks.median())
print(marks.mode())
print(marks.min()) #or max()

# unique values
print(marks.unique())

# No. of unique values
print(marks.nunique())

# Freq. of each value
print(marks.value_counts())

# Sort by  values
print(marks.sort_values())

# Sort by index
print(marks.sort_index())

# Detect missing values
print(marks.isnull())

# Detect non missing value
print(marks.notnull())

# Drop missing values
print(marks.dropna())

# Convert data type
print(marks.astype(float))

# Create an independent copy
print(marks.copy())

# Sort by descending
print(marks.sort_values(ascending=False))

# Sort by ascending
print(marks.sort_values(ascending=True))
```
